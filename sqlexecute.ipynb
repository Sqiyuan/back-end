```python
# 职位数据管理数据库操作类   DAO：database access object
class UserDao(BaseDao):

    # user info
    def createUser(self, params=[]):
        id_res = list(self.execute('SELECT id FROM user ORDER BY id DESC LIMIT 1;'))

        sql = "insert into user (id, username, password, uni_id, name, stu_num, pho_num) " \
              "values (%d, %s, %s, %d, %s, %s, %s)"  # %s是占位符
        params = id_res + params
        result = self.execute(sql ,params)
        self.commit()
        return result
    
    def usernameMatch(self, username):
        sql = '''SELECT * FROM `user` where username='{}';'''.format(username)
        self.execute(sql)
        resultSet = self.fetchall()
        return resultSet
```
